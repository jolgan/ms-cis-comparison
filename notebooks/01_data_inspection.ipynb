# Data Inspection: Lithuanian and Mexican CIS Cohorts

CIS (Clinically Isolated Syndrome) is the first neurological episode that may indicate Multiple Sclerosis - but where a formal MS diagnosis cannot yet be made.

## 1. Load Libraries and Data

Import pandas library and load both Excel files into memory as dataframes for inspection.

```python
import pandas as pd

lt_df = pd.read_excel('../data/raw/lithuanian_cis.xlsx')
mx_df = pd.read_excel('../data/raw/mexican_cis.xlsx')

print('Files loaded successfully.')
```

## 2. Dataset Shapes

Shape is basically how big a dataset is - specifically, how many rows and columns. Here, pandas prints the shape of both dataframes. Each row represents a patient who presented with Clinically Isolated Syndrome (CIS), and each column represents a variable (patient information e.g. age, sex, presence of lesions in MRI).

```python
print('Lithuanian CIS shape (rows, columns):', lt_df.shape)
print('Mexican CIS shape    (rows, columns):', mx_df.shape)
```

## 3. Column Names

This step lists all column names for each dataset. It displays the variables in the Lithuanian study, and reveals that the Mexican dataset has a lot of unnamed columns. This could indicate blank rows above the real headers. This is checked in the next step.

```python
print('=== Lithuanian CIS columns ===')
for col in lt_df.columns:
    print(' ', col)

print()
print('=== Mexican CIS columns ===')
for col in mx_df.columns:
    print(' ', col)
```

## 4. First 5 Rows

The **.head()** method pulls the first five rows of each dataset to inspect their actual values, data types, and formatting. As shown below, the Lithuanian dataset is properly structured, but the Mexican dataset reveals that the column headers actually only start on row 3.

```python
print('=== Lithuanian CIS - first 5 rows ===')
display(lt_df.head())
```

Pandas uses the term **NaN** interchangeably with **null** - it stands for Not a Number and represents a missing value or empty cell. Sometimes Excel files are formatted for human reading rather than machine reading, which could explain why there are 2 rows of these in the Mexican dataset. This needs to be fixed, as **pd.read_excel()** assumes by default that the very first row contains the column headers.

```python
print('=== Mexican CIS - first 5 rows ===')
display(mx_df.head())
```

## 5. Null Value Counts per Column

Nulls are not zero. Zero is a value, indicating something was measured and the result was zero. A null means nothing was recorded at all. Null values do not contribute meaningfully to analysis, so this step counts the missing values in each column before any analysis takes place. The output displays only columns with at least one null.

```python
print('=== Lithuanian CIS - null counts per column ===')
lt_nulls = lt_df.isnull().sum()
display(lt_nulls[lt_nulls > 0].rename('null_count').to_frame() if lt_nulls.any() else 'No nulls found.')
print(f'Total nulls: {lt_nulls.sum()}')
```

```python
print('=== Mexican CIS - null counts per column ===')
mx_nulls = mx_df.isnull().sum()
display(mx_nulls[mx_nulls > 0].rename('null_count').to_frame() if mx_nulls.any() else 'No nulls found.')
print(f'Total nulls: {mx_nulls.sum()}')
```

## Data Cleaning

This section addresses structural issues identified during inspection: blank header rows in the Mexican dataset, a ghost summary row in the Lithuanian dataset, numeric-only column names that obscure their meaning, and columns too sparse to be analytically useful.

### 6.1 Mexican Dataset: Fix Column Headers

The file loaded with `Unnamed: 0` through `Unnamed: 19` because the first row of the Excel file is blank, so pandas auto-generated column names. Reload with `header=None` to inspect the raw rows and confirm which row holds the actual column names, then reload with the correct `header` argument.

```python
# Reload with no header inference to inspect the raw row structure
mx_raw = pd.read_excel('../data/raw/mexican_cis.xlsx', header=None)
print(f'Raw file shape (no header consumed): {mx_raw.shape}')
print()
print('First 5 rows (row 3 should contain the real column names):')
display(mx_raw.head(5))
```

```python
# Row 3 (0-indexed from the start of the file) contains the actual column names;
# rows 0-2 are the blank rows and old unnamed header, skipped automatically.
mx_df = pd.read_excel('../data/raw/mexican_cis.xlsx', header=3)
print('Corrected Mexican CIS column names:')
for col in mx_df.columns:
    print(' ', col)
print()
print(f'Shape after header fix: {mx_df.shape}')
```

### 6.2 Lithuanian Dataset: Remove Ghost Row

Inspection showed exactly 1 null in nearly every column — the signature of a single empty or summary row. The patient-ID column `Nr` is the clearest signal: any row where `Nr` is null is not a genuine patient record. Drop it and confirm the row count drops from 139 to 138.

```python
ghost_mask = lt_df['Nr'].isna()
print(f'Rows with null Nr (patient ID): {ghost_mask.sum()}')
print('Ghost row:')
display(lt_df[ghost_mask])

lt_df = lt_df[~ghost_mask].reset_index(drop=True)
print(f'Row count: 139 -> {len(lt_df)}')
assert len(lt_df) == 138, 'Expected 138 rows after removing ghost row'
print('Confirmed: 138 rows.')
```

### 6.3 Both Datasets: Rename Numeric Columns and Drop Sparse Columns

The Lithuanian dataset has columns named `3`, `4`, `5`, `6`, `8` — these are cranial nerve assessments (CN3, CN4, CN5, CN6, CN8). Rename them to make the data self-documenting. Then, for both datasets, drop any column where more than 50% of values are null, as those columns carry too little signal to be analytically useful.

```python
# Rename numeric-only cranial nerve columns in the Lithuanian dataset
lt_df = lt_df.rename(columns={'3': 'CN3', '4': 'CN4', '5': 'CN5', '6': 'CN6', '8': 'CN8'})
print('Renamed: 3->CN3, 4->CN4, 5->CN5, 6->CN6, 8->CN8')

def drop_sparse(df, threshold=0.5):
    null_frac = df.isnull().mean()
    sparse = null_frac[null_frac > threshold].index.tolist()
    if sparse:
        print(f'  Dropping {len(sparse)} column(s) with >{int(threshold * 100)}% nulls: {sparse}')
    else:
        print(f'  No columns exceed the {int(threshold * 100)}% null threshold.')
    return df.drop(columns=sparse)

print()
print('Lithuanian CIS:')
lt_df = drop_sparse(lt_df)
print(f'  Final shape: {lt_df.shape}')

print()
print('Mexican CIS:')
mx_df = drop_sparse(mx_df)
print(f'  Final shape: {mx_df.shape}')
```

### 6.4 Export Cleaned Datasets

Save both cleaned DataFrames as CSV files to `data/cleaned/`. The DataFrame index is excluded since both datasets already contain patient-ID columns.

```python
import os
os.makedirs('../data/cleaned', exist_ok=True)

lt_df.to_csv('../data/cleaned/lithuanian_cleaned.csv', index=False)
mx_df.to_csv('../data/cleaned/mexican_cleaned.csv', index=False)

print('Exported:')
print(f'  data/cleaned/lithuanian_cleaned.csv  -- {lt_df.shape[0]} rows x {lt_df.shape[1]} columns')
print(f'  data/cleaned/mexican_cleaned.csv     -- {mx_df.shape[0]} rows x {mx_df.shape[1]} columns')
```
